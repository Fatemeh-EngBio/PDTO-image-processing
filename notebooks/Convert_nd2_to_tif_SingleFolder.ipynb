```python
import numpy as np
import matplotlib.pyplot as plt
import mahotas as mh
import imutils
import nd2
import cv2
import os
pwd = os.getcwd()
```

```python
folder_address = ['./raw_images']
```

```python
all_imgs_names = [f for f in os.listdir(folder_address[0]) \
                  if (os.path.splitext(os.path.join(pwd,f))[1] == '.nd2')]

print(len(all_imgs_names))
```

```python
for img_num in range(len(all_imgs_names)):
    print(all_imgs_names[img_num])

    img_array = nd2.imread(os.path.join(folder_address[0], all_imgs_names[img_num])) # read to numpy array

    image_name = all_imgs_names[img_num]
    
    plt.figure()
    plt.imshow(img_array, 'gray')
    plt.show()

    cv2.imwrite(os.path.join(folder_address[0], os.path.splitext(os.path.basename(os.path.join(folder_address[0], 
                                                                                               all_imgs_names[img_num])))[0]+'.tif'), img_array)
```
